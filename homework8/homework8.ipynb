```python
#6
median_latitude = df['latitude'].median()
print(median_latitude)
median_longitude = df['longitude'].median()
print(median_longitude)
```

```python
#7.
df['latitude_distance'] = (df['latitude'] - median_latitude).abs()
print(df)
```

```python
#8
df['longitude_distance'] = (df['longitude'] - median_longitude).abs()
print(df)
```

```python
#9
print(df.shape)
```

```python
#10
closest_lat_idx = df['latitude_distance'].idxmin()
closest_lat_state = df.loc[closest_lat_idx, 'state'] 
closest_lat_val = df.loc[closest_lat_idx, 'latitude_distance']
print(f"{closest_lat_state} by {closest_lat_val:.2f}")

farthest_lat_idx = df['latitude_distance'].idxmax()
farthest_lat_state = df.loc[farthest_lat_idx, 'state'] 
farthest_lat_val = df.loc[farthest_lat_idx, 'latitude_distance']
print(f"{farthest_lat_state} by {farthest_lat_val:.2f}")

closest_lon_idx = df['longitude_distance'].idxmin()
closest_lon_state = df.loc[closest_lon_idx, 'state'] 
closest_lon_val = df.loc[closest_lon_idx, 'longitude_distance']
print(f"{closest_lon_state} by {closest_lon_val:.2f}")

farthest_lon_idx = df['longitude_distance'].idxmax()
farthest_lon_state = df.loc[farthest_lon_idx, 'state'] 
farthest_lon_val = df.loc[farthest_lon_idx, 'longitude_distance']
print(f"{farthest_lon_state} by {farthest_lon_val:.2f}")
```

```python
import seaborn as sns
```

```python
titanic = sns.load_dataset('titanic')
print(titanic)
```

```python
titanic_cleaned = titanic.dropna(axis=1)
print(titanic_cleaned)
```

```python
print(titanic_cleaned.shape)
```

```python
print(titanic_cleaned.head())
print(titanic_cleaned.tail())
```

```python
print(titanic_cleaned.iloc[:,0]) #survive = survived the accident
print(titanic_cleaned.iloc[:,1]) #pclass = poverty class 
print(titanic_cleaned.iloc[:,2]) #sex = gender
print(titanic_cleaned.iloc[:,3]) #sibsp = represents the number of siblings or spouses a passenger had aboard
print(titanic_cleaned.iloc[:,4]) #parch = represents the number of  parents or children a passenger was 
                                        #traveling with aboard the ship
print(titanic_cleaned.iloc[:,5]) #fare = how much money spent on ticket
print(titanic_cleaned.iloc[:,6]) #class = first/second/third class
print(titanic_cleaned.iloc[:,7]) # who = woman or man
print(titanic_cleaned.iloc[:,8]) #adult male = is it a male true or false
print(titanic_cleaned.iloc[:,9]) # alive = did they survive
print(titanic_cleaned.iloc[:,10]) # were they alone (no siblins or parents or friends with them)
```

```python
import matplotlib.pyplot as plt
```

```python
x = titanic_cleaned['sibsp']
y = titanic_cleaned['parch']
a = titanic_cleaned['survived']
b = titanic_cleaned['pclass']
c= titanic_cleaned['fare']

fig, axes = plt.subplots(2, 2, figsize=(12, 10))
#for top left
axes[0, 0].scatter(x, y, color='darkorange', label='Family Units')
axes[0, 0].set_title('Family Size: Siblings vs. Parents')
axes[0, 0].set_xlabel('Siblings/Spouses (Count)')
axes[0, 0].set_ylabel('Parents/Children (Count)')
axes[0, 0].legend() 

#for top right 
axes[0, 1].scatter(a, b, color='green')
axes[0, 1].set_title('Survival vs. Passenger Class')
axes[0, 1].set_xlabel('Survived (1=Survived, 0=Deceased)')
axes[0, 1].set_ylabel('Passenger class (1 = first, 2 = second, 3 = third')

#for bottom left
axes[1, 0].hist(b, bins=6, color='crimson')
axes[1, 0].set_title("Passenger Class Distribution")
axes[1, 0].set_xlabel("Class Type")
axes[1, 0].set_ylabel("Passenger Count")

#For bottom right 
sns.boxplot(ax=axes[1, 1], data=titanic_cleaned, x = 'pclass', y='fare') 
axes[1, 1].set_title('Passenger Class Vs Fare')
axes[1, 1].set_xlabel('Passenger Class (1st, 2nd, 3rd)')
axes[1, 1].set_ylabel('Fare (British Pounds)') 
axes[1, 1].set_ylim(0, 100)


plt.tight_layout()
plt.show()
```

```python
#top right: This compares survivors with passanger class. These were chosen to see if 
#          social standing influenced the chance of survival. From this you can tell that there were survivors
#           and deaths among all classes.
#Top left: This plot uses sibsp (siblings/spouses) and parch (parents/children). I chose these because 
#           they are the only two columns that describe family relationships on board. From this you can tell
# Bottom left: this represents the passanger class column. I chose this to understand the 
# population of the ship and how many people were in each class level. From this we can see that a majority of 
# passangers were in the third class. 
#Bottom right: This was pciked to show the relationship between the ticket prices and the class ration. From this 
#     we can see that dare prices were more expensive for first class and relatively cheapter at a tighther range
#     for third class.
```
